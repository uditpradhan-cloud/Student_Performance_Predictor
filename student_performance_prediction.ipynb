```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "model__n_estimators": [100, 200, 300],
    "model__max_depth": [None, 5, 10, 15],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4]
}

# In simple terms:
# It tells GridSearchCV:
# "Take my training data and find the best Random Forest settings from all the combinations I gave you."

# For example:
# n_estimators → number of trees
# max_depth → maximum depth of each tree
# min_samples_split → minimum samples needed to split a node
# min_samples_leaf → minimum samples allowed in a leaf

#What does this mean : 
# 100 trees + depth 5 + split 2 + leaf 1
# 100 trees + depth 5 + split 2 + leaf 2
# 100 trees + depth 5 + split 2 + leaf 4
# ...
# 200 trees + depth 10 + split 5 + leaf 2
# ...
# 300 trees + depth 15 + split 10 + leaf 4

#Totol 108 Combinations
```

```python
grid_search = GridSearchCV(
    rf_model,
    param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)
```

```python
zero_classifier = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced"
    ))
])
```

```python
zero_classifier.fit(X_train, y_zero_train)
```

```python
zero_predictions = zero_classifier.predict(X_test)
```

```python
nonzero_train_indices = X_train.index[y_train > 0]

X_nonzero_train = X_train.loc[nonzero_train_indices]
y_nonzero_train = y_train.loc[nonzero_train_indices]
```

```python
nonzero_rf = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=300,
        random_state=42
    ))
])
```

```python
nonzero_rf.fit(X_nonzero_train, y_nonzero_train)
```

```python
nonzero_predictions = nonzero_rf.predict(X_test)
```

```python
final_predictions_two_stage = np.where(
    zero_predictions == 1,
    0,
    nonzero_predictions
)
```

```python
mae_two_stage = mean_absolute_error(
    y_test,
    final_predictions_two_stage
)

rmse_two_stage = np.sqrt(
    mean_squared_error(
        y_test,
        final_predictions_two_stage
    )
)

r2_two_stage = r2_score(
    y_test,
    final_predictions_two_stage
)

print("Two-Stage Model")
print("MAE:", mae_two_stage)
print("RMSE:", rmse_two_stage)
print("R²:", r2_two_stage)
```

```python
error_two_stage = np.abs(
    final_predictions_two_stage - y_test
)

print(
    "Predictions within ±2 grades:",
    (error_two_stage <= 2).mean() * 100,
    "%"
)
```

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_mae = []
cv_rmse = []
cv_r2 = []
cv_within2 = []

for train_idx, test_idx in skf.split(X, y_zero):

    X_cv_train = X.iloc[train_idx]
    X_cv_test = X.iloc[test_idx]

    y_cv_train = y.iloc[train_idx]
    y_cv_test = y.iloc[test_idx]

    y_zero_cv_train = y_zero.iloc[train_idx]

    # Stage 1: zero-grade classifier
    classifier = Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced"
        ))
    ])

    classifier.fit(X_cv_train, y_zero_cv_train)

    zero_pred = classifier.predict(X_cv_test)

    # Stage 2: regression for non-zero grades
    nonzero_mask = y_cv_train > 0

    regression = Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=300,
            random_state=42
        ))
    ])

    regression.fit(
        X_cv_train.loc[nonzero_mask],
        y_cv_train.loc[nonzero_mask]
    )

    grade_pred = regression.predict(X_cv_test)

    # Combine
    final_pred = np.where(
        zero_pred == 1,
        0,
        grade_pred
    )

    # Metrics
    error = np.abs(final_pred - y_cv_test)

    cv_mae.append(mean_absolute_error(y_cv_test, final_pred))
    cv_rmse.append(np.sqrt(mean_squared_error(y_cv_test, final_pred)))
    cv_r2.append(r2_score(y_cv_test, final_pred))
    cv_within2.append((error <= 2).mean() * 100)


print("Two-Stage 5-Fold Cross-Validation")
print("-----------------------------------")

print("MAE scores:", cv_mae)
print("Mean MAE:", np.mean(cv_mae))

print("\nRMSE scores:", cv_rmse)
print("Mean RMSE:", np.mean(cv_rmse))

print("\nR² scores:", cv_r2)
print("Mean R²:", np.mean(cv_r2))

print("\nWithin ±2 scores:", cv_within2)
print("Mean Within ±2:", np.mean(cv_within2), "%")
```

```python
original_cv_mae = []
original_cv_rmse = []
original_cv_r2 = []
original_cv_within2 = []

for train_idx, test_idx in skf.split(X, y_zero):

    X_cv_train = X.iloc[train_idx]
    X_cv_test = X.iloc[test_idx]

    y_cv_train = y.iloc[train_idx]
    y_cv_test = y.iloc[test_idx]

    model_cv = Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=200,
            random_state=42
        ))
    ])

    model_cv.fit(X_cv_train, y_cv_train)

    pred_cv = model_cv.predict(X_cv_test)

    error_cv = np.abs(pred_cv - y_cv_test)

    original_cv_mae.append(
        mean_absolute_error(y_cv_test, pred_cv)
    )

    original_cv_rmse.append(
        np.sqrt(mean_squared_error(y_cv_test, pred_cv))
    )

    original_cv_r2.append(
        r2_score(y_cv_test, pred_cv)
    )

    original_cv_within2.append(
        (error_cv <= 2).mean() * 100
    )


print("Original Random Forest - 5 Fold Cross-Validation")
print("-----------------------------------------------")

print("MAE scores:", original_cv_mae)
print("Mean MAE:", np.mean(original_cv_mae))

print("\nRMSE scores:", original_cv_rmse)
print("Mean RMSE:", np.mean(original_cv_rmse))

print("\nR² scores:", original_cv_r2)
print("Mean R²:", np.mean(original_cv_r2))

print("\nWithin ±2 scores:", original_cv_within2)
print("Mean Within ±2:", np.mean(original_cv_within2), "%")
```
